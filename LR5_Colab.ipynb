## Крок 4. Подивитися результати та графіки

```python
import json
from pathlib import Path
from IPython.display import display, Image, Markdown
metrics = json.loads(Path('results/metrics.json').read_text(encoding='utf-8'))
for key in ['corpus_source', 'total_tokens', 'unique_words', 'test_count',
            'correct_top1','accuracy_top1', 'accuracy_top3',
            'accuracy_one_edit', 'accuracy_two_edits', 'avg_correction_ms']:
    print(f'{key}: {metrics.get(key)}')
for name in ['accuracy.png', 'frequency.png', 'edit_distance.png']:
    display(Image(filename=str(Path('results') / name), width=730))
```

## Крок 5. Завантажити архів результатів
Архів міститиме код, використаний корпус `data/shakespeare.txt`, звіт, таблиці, графіки і файли сайту. Для фінального GitHub-архіву додайте окремо цей блокнот `LR5_Colab.ipynb` із початкового ZIP.


```python
from pathlib import Path
from zipfile import ZipFile, ZIP_DEFLATED
from google.colab import files
out_name = 'LR5_results_for_GitHub.zip'
with ZipFile(out_name, 'w', ZIP_DEFLATED) as z:
    for base in ['lab5.py', 'README.md', 'requirements.txt', '.gitignore']:
        z.write(base, base)
    for dirname in ['tests', 'results', 'docs', 'data']:
        for p in sorted(Path(dirname).rglob('*')):
            if p.is_file() and '__pycache__' not in str(p) and p.suffix != '.pyc':
                z.write(p, str(p))
print('Архів створено:', Path(out_name).stat().st_size, 'байт')
files.download(out_name)
```
